```python
# Carga del dataset

import pandas as pd

# leer los datos directamente desde el zip

dataframe = pd.read_csv(
    "../data/sentences.csv.zip",
    index_col=False,
    compression="zip",
)
```

Ahora haremos la transforamción numérica por medio de la matriz término-documentos, matriz de TF-IDF 

```python
# Particionamiento de los datos

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    dataframe.phrase,
    dataframe.target,
    test_size=0.3,
    shuffle=False,
)
```

```python
# Construcción de la matriz documento-término

from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(
    lowercase=True, # poner todo en minúscula
    analyzer="word", # analizar palabras
    token_pattern=r"\b[a-zA-Z]\w+\b", # solo acepta palabras
    stop_words="english", # idioma ingles
    max_df=0.99, # si una palabra aparece el 90% de las veces
    # entonces no se tienen en cuenta porque son stopwords
    min_df=2, # frecuencia mínima para usar palabras
    binary=True, # binaria es la forma de indicar que esta
    # o no está
)

# aplicar todo el countvectorizer
# todas las funcionalidades, es básicamnete
# crea la matriz término documento

vectorizer.fit(X_train)

# ver las palabras
vectorizer.get_feature_names_out()
```

Nota: es importante entender que las mayúsculas y minúsculas hacen palabras diferentes, por ello es que hay que hacer un pre-procesamiento. 

```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression(max_iter=1000)

X_train_vectorized = vectorizer.transform(X_train)

clf.fit(
    X_train_vectorized,
    y_train,
)
```

```python
# Evaluación del modelo

from sklearn.metrics import accuracy_score

# Muestra de entrenamiento:

accuracy_score(
    y_true=y_train,
    y_pred=clf.predict(X_train_vectorized),
)
```

```python
# Muestra de prueba:

X_test_vectorized = vectorizer.transform(X_test)
predictions = clf.predict(X_test_vectorized)

accuracy_score(
    y_true=y_test,
    y_pred=predictions,
)
```

A partir de lo anterior nos damos cuenta que para el entrenamiento tenemos un buen accuracy, pero lo que pasa es que en test ya tenemos un mal accuracy. Esto es debeido a que las clases están desbalanceadas y por ende se tendrán clasificaciones sesgadas. 

Esto es lo que sucede en casos de clasficación cuando las categorías que se van a predecir en el test de entramiento estan muy desbalalanceadas. Esto puede verse con value_counts()

```python
from sklearn.metrics import ConfusionMatrixDisplay

disp = ConfusionMatrixDisplay.from_predictions(
    y_test,
    predictions,
    cmap="Greens",
)
disp.figure_.suptitle("Confusion Matrix")
print(f"Confusion matrix:\n{disp.confusion_matrix}")
```

```python
import pickle

with open("../submission/clf.pkl", "wb") as file:
    pickle.dump(clf, file)

with open("../submission/vectorizer.pkl", "wb") as file:
    pickle.dump(vectorizer, file)
```

```python
  # Uso del modelo

with open("../submission/clf.pkl", "rb") as file:
    new_clf = pickle.load(file)

with open("../submission/vectorizer.pkl", "rb") as file:
    new_vectorizer = pickle.load(file)

accuracy_score(
    y_true=dataframe.target,
    y_pred=new_clf.predict(new_vectorizer.transform(dataframe.phrase)),
)
```
